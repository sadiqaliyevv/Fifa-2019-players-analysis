## Preprocessing (Oyunçu dəyərinin rəqəmə çevrilməsi)

##### Value sutununa baxmaq isteyirem. Sutunun ilk 10 reqemini nece gore bilerem?

```python
df['Value'].head(10)
```

##### Burda Euro isaresini cixaririq, M herfini 1000000-la, K herfini ise 1000-le evezleyirik

```python
def convert_value(val):
    val = str(val) 
    val = val.replace('€', '') 
    if 'M' in val:
        return float(val.replace('M', '')) * 1_000_000
    elif 'K' in val:
        return float(val.replace('K', '')) * 1_000
    else:
        return float(val)


df['Value'] = df['Value'].apply(convert_value)
df['Value'].head(10)
```

 İndi isə sütunu iki hissəyə böləcək kod yazın və bu iki sütunu bir birinə vuraraq real "Value" sütununu əldə edin (float tipində)

```python
df[['Value_num', 'Value_unit']] = df['Value'].str.replace('€', '', regex=False).str.extract(r'([\d\.]+)([MK]?)')
df['Value_num'] = df['Value_num'].astype(float)

df['Multiplier'] = df['Value_unit'].map({'M': 1_000_000, 'K': 1_000, '': 1})

df['Value_real'] = df['Value_num'] * df['Multiplier']

```

##### Menim sutunumda null deyerler ola biler. Men isteyirem ki bunu 0 ile evez edim

```python
df['Value_real'] = df['Value_real'].fillna(0)
```

## Visualization and Analysis

#####  Oyunçuların Pozisiya paylanması (Categorical)

Bir fiqur (10,10) yaradın Bu fiqur daxilində hansı pozisiyadan neçə nəfər uyğun chart vasitəsilə göstərin ("Position" sütunu)

```python
position_counts = df['Position'].value_counts()
plt.figure(figsize=(10, 10))
sns.barplot(x=position_counts.values, y=position_counts.index, palette="viridis")

plt.title('Oyunçuların Pozisiya Paylanması', fontsize=16)
plt.xlabel('Oyunçu sayı', fontsize=10)
plt.ylabel('Pozisiya', fontsize=10)
plt.tight_layout()
plt.show()
```

#### Oyunçuların yaş paylanması (Numeric)

Bir fiqur (10,10) yaradın Bu fiqur daxilində hansı yaş sütununu uyğun chart vasitəsilə göstərin ("Age" sütunu)

```python
plt.figure(figsize=(10, 10))
sns.histplot(df['Age'], bins=15, kde=False, color='skyblue', edgecolor='black')

plt.title('Oyunçuların Yaş Paylanması', fontsize=16)
plt.xlabel('Yaş', fontsize=12)
plt.ylabel('Oyunçu sayı', fontsize=12)

plt.tight_layout()
plt.show()
```

Bir də bunu kde plot vasitəsilə verin

```python
plt.figure(figsize=(5, 5))
sns.kdeplot(data=df, x='Age',fill=True, color='purple')
plt.title('Oyunçuların Yaş Sıxlıq Qrafiki')
plt.xlabel('Yaş')
plt.ylabel('Sıxlıq')
plt.tight_layout()
plt.show()
```

### Oyunçuların Potensialları və Gücləri arasında əlaqə

Buna birinci seabornda olan lineplotla baxmaq isteyirem

```python
plt.figure(figsize=(9, 6))
sns.lineplot(data=df,x="Overall",y="Potential",ci=None)
plt.title("Oyunçuların Potensialları və Gücləri arasında əlaqə")
plt.xlabel("Güc")
plt.ylabel("Potensial")
plt.grid(True)
plt.tight_layout()
plt.show()
```

##### Oyunçuların Potensialları və Gücləri-İstifadə olunan ayaq arasında

```python
avg_pot_foot = df.groupby(["Overall", "Preferred Foot"])["Potential"].mean().reset_index()
plt.figure(figsize=(12, 6))
sns.lineplot(data=avg_pot_foot,x="Overall",y="Potential",hue="Preferred Foot",marker="o")

plt.title("Oyunçuların Potensialları və Gücləri - İstifadə olunan ayağa görə")
plt.xlabel("Güc")
plt.ylabel("Orta Potensial")
plt.legend(title="İstifadə olunan ayaq")
plt.grid(True)
plt.tight_layout()
plt.show()
```
